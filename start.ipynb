```python
import os

structure = {
    "backend": {
        "app": {
            "api": [
                "__init__.py",
                "generate.py",
                "health.py"
            ],
            "pipeline": [
                "__init__.py",
                "intent_extractor.py",
                "system_designer.py",
                "schema_generator.py",
                "master_pipeline.py",
                "prompts.py"
            ],
            "schemas": [
                "__init__.py",
                "intent_schema.py",
                "architecture_schema.py",
                "ui_schema.py",
                "api_schema.py",
                "db_schema.py",
                "auth_schema.py",
                "validation_schema.py",
                "response_schema.py"
            ],
            "validators": [
                "__init__.py",
                "json_validator.py",
                "schema_validator.py",
                "consistency_validator.py",
                "business_validator.py"
            ],
            "repair": [
                "__init__.py",
                "repair_engine.py",
                "targeted_regenerator.py",
                "retry_handler.py"
            ],
            "runtime": [
                "__init__.py",
                "renderer.py",
                "component_mapper.py"
            ],
            "services": [
                "__init__.py",
                "openai_service.py",
                "logging_service.py"
            ],
            "evaluation": [
                "__init__.py",
                "test_cases.py",
                "metrics.py",
                "benchmark_runner.py"
            ],
            "utils": [
                "__init__.py",
                "json_utils.py",
                "parser_utils.py",
                "constants.py",
                "helpers.py"
            ],
            "files": [
                "config.py",
                "main.py"
            ]
        },
        "tests": [
            "test_pipeline.py",
            "test_validation.py",
            "test_repair.py",
            "test_runtime.py"
        ],
        "files": [
            ".env",
            "requirements.txt",
            "run.py",
            "README.md"
        ]
    },

    "frontend": {
        "public": [
            "favicon.ico"
        ],
        "src": {
            "components": [
                "PromptBox.jsx",
                "GenerateButton.jsx",
                "OutputCard.jsx",
                "ValidationPanel.jsx",
                "RepairLogs.jsx",
                "Loader.jsx",
                "Navbar.jsx"
            ],
            "pages": [
                "Home.jsx"
            ],
            "services": [
                "api.js"
            ],
            "renderer": [
                "DynamicRenderer.jsx",
                "ComponentFactory.jsx"
            ],
            "utils": [
                "formatters.js",
                "constants.js"
            ],
            "files": [
                "App.jsx",
                "main.jsx",
                "index.css",
                "App.css"
            ]
        },
        "files": [
            "package.json",
            "vite.config.js",
            ".gitignore",
            "README.md"
        ]
    },

    "docs": {
        "screenshots": [],
        "files": [
            "architecture.png",
            "workflow.png",
            "demo-notes.md"
        ]
    },

    "root_files": [
        ".gitignore",
        "README.md",
        "LICENSE"
    ]
}


def create_structure(base_path, structure):
    for name, content in structure.items():

        if name == "files" or name == "root_files":
            for file in content:
                file_path = os.path.join(base_path, file)
                open(file_path, "w").close()

        elif isinstance(content, dict):
            folder_path = os.path.join(base_path, name)
            os.makedirs(folder_path, exist_ok=True)
            create_structure(folder_path, content)

        elif isinstance(content, list):
            folder_path = os.path.join(base_path, name)
            os.makedirs(folder_path, exist_ok=True)

            for file in content:
                file_path = os.path.join(folder_path, file)
                open(file_path, "w").close()


create_structure(".", structure)

print("✅ Complete AI App Compiler project structure created successfully!")
```
